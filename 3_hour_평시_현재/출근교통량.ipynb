```python
import pandas as pd
import numpy as np
from datetime import datetime, date
```

```python
df = pd.read_excel("교차로 접근로별 교통량_이마트.xlsx", sheet_name="2303")
df
```

```python
df['시간'] = pd.to_datetime(df['시간'])
```

```python
df['요일'] = df['시간'].dt.dayofweek
```

```python
filtered_df = df[(df['요일'] != 5) & (df['요일'] != 6)]
df = filtered_df
```

```python
df_W1 = df[df['접근로명']=='보도2교방향']
df_W1 = df_W1.reset_index(drop=True)
```

```python
df_W1['출근시간'] = df_W1['시간'].dt.strftime('%H:%M:%S')
```

```python
df_W1.drop(columns=['시간'], inplace=True)
```

```python
six = df_W1[(df_W1['출근시간'] == '17:00:00')]
seven = df_W1[(df_W1['출근시간'] == '18:00:00')]
eight = df_W1[(df_W1['출근시간'] == '19:00:00')]
```

```python
max_six = six['교통량'].max()
max_seven = seven['교통량'].max()
max_eight = eight['교통량'].max()
max_six, max_seven, max_eight
```

```python
mean_six = six['교통량'].mean()
mean_seven = seven['교통량'].mean()
mean_eight = eight['교통량'].mean()
mean_six, mean_seven, mean_eight
```

```python
max_six + max_seven + max_eight
```
