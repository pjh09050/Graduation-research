```python
import numpy as np
a = [207.78, 208.973, 212.75, 211.385, 213.165, 213.157, 212.04, 210.883, 212.967, 207.059]

print('최소 :', np.round(np.min(a),1))
print('최대 :', np.round(np.max(a),1))
print('평균 :', np.round(np.mean(a),1))
print('분산:', np.round(np.var(a),1))
print('표준편차:', round(np.std(a),1))
```
