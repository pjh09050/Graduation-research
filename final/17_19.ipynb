## W1 지수분포에 따른 차량 등장 시간

```python
import numpy as np
import random
W1_arrival_rate = 1500 / 3600  # 차량 도착율 계산
W1_time_list = [] # W1 등장 차량 시간 리스트
W1_arrival_intervals = np.random.exponential(scale=1/W1_arrival_rate, size=1500)
cumulative_interval = 0
for i, interval in enumerate(W1_arrival_intervals):
    # W1출발부터 모든 경로에 대한 확률 가중치 부여 후 경로 선택
    car_direction = random.choices(population=["routeW1_N1", "routeW1_S1", "routeW1_N2", "routeW1_S2", "routeW1_S3", "routeW1_E3"], weights=[0.1,0.1,0.04,0.04,0.072,0.648], k=1)[0]
    cumulative_interval += interval
    if cumulative_interval > 3600:
        break
    if car_direction == "routeW1_N1":
        departLane = 3
    elif car_direction == "routeW1_S1":
        departLane= 0
    elif car_direction == "routeW1_N2":
        departLane = random.randint(1,2)
    elif car_direction == "routeW1_S2":
        departLane = random.randint(1,2)
    elif car_direction == "routeW1_S3":
        departLane = random.randint(1,2)
    elif car_direction == "routeW1_E3":
        departLane = random.randint(1,2)
    
    W1_time_list.append((car_direction, np.round(cumulative_interval,1), departLane))
#print(W1_time_list)
print(len(W1_time_list), '대 차량 생성')

for i in range(len(W1_time_list)):
    print('    <vehicle id="route_%i" type="typeCAR" route="%s" depart="%.1f" departLane="%i" />' % (i, W1_time_list[i][0],W1_time_list[i][1], W1_time_list[i][2]))
```

## N1 지수분포에 따른 차량 등장 시간

```python
N1_arrival_rate = 455 / 3600  # 차량 도착율 계산
N1_time_list = [] # N1 차량 등장 시간 리스트
N1_arrival_intervals = np.random.exponential(scale=1/N1_arrival_rate, size=455)
cumulative_interval = 0
for i, interval in enumerate(N1_arrival_intervals):
    cumulative_interval += interval
    if cumulative_interval > 3600:
        break
    N1_time_list.append(np.round(cumulative_interval,1))
print(N1_time_list)
print(len(N1_time_list))
```
