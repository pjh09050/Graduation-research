```python
import random

total_cars = 1477
cars_in_east = 0
cars_in_south = 0
cars_in_north = 0
cars = 0
arrival_times = random.sample(range(3600), total_cars)
arrival_times.sort()

for i in range(3600):
    if i in arrival_times:
        car_direction = random.choices(population=["N", "S", "E"], weights=[0.1, 0.1, 0.8], k=1)[0]
        cars = cars + 1
        print(f'{i}초에 "{car_direction}"차량이 등장합니다.')
        
        if car_direction == "E":
            cars_in_east += 1
        elif car_direction == "S":
            cars_in_south += 1
        elif car_direction == "N":
            cars_in_north += 1
    else:
        car_direction = ""
        
print("E방향으로 이동한 차량 수:", cars_in_east)
print("S방향으로 이동한 차량 수:", cars_in_south)
print("N방향으로 이동한 차량 수:", cars_in_north)
print("등장한 차량 수:", cars)
```
